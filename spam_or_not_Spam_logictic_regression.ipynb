```python
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('/content/mail_data.csv')
```

```python
df.head()
```

```python
#replace the null values with null string
mail_data = df.where((pd.notnull(df)),'')
```

```python
mail_data.head()
```

```python
mail_data.describe()
```

```python
mail_data.columns
```

```python
mail_data.info()
```

```python
mail_data.duplicated().sum() ## duplicate
```

label encoding


```python
# label spam as 0 , ham as 1
mail_data.loc[mail_data['Category'] == 'spam' , 'Category', ] = 0
mail_data.loc[mail_data['Category'] == 'ham' , 'Category', ] = 1
```

```python
mail_data.head()
```

```python
#seperating data as text and label
X = mail_data['Message']
Y = mail_data['Category']
```

```python
#splitting data into train test data
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=3)
```

```python
print(X.shape, X_train.shape, X_test.shape)
print(Y.shape, Y_train.shape, Y_test.shape)
```

**Feature extraction**
from text is the process of converting unstructured text (sentences, documents, reviews, emails, etc.) into numerical features that a machine learning model can understand.

Since ML algorithms work only with numbers, text must first be transformed into vectors.

Why is it needed?

For example, consider two movie reviews:

"This movie is amazing."
"This movie is terrible."

A machine cannot understand the words "amazing" or "terrible" directly. Feature extraction converts them into numerical representations.


# **TF-IDF (Term Frequency-Inverse Document Frequency)**
This is the most commonly used feature extraction technique in classical machine learning.

Instead of simply counting words, TF-IDF gives higher importance to unique words and lower importance to very common words.

Formula
Term Frequency (TF)

Measures how often a word appears in a document.

TF=
Total number of words in the document
Number of occurrences of a word
	​

Inverse Document Frequency (IDF)

Measures how unique a word is across all documents.

IDF=log( Documents containing the word Total documents )
TF−IDF=TF×IDF

```python
#transform text to feature vector / numerical values
feature_extraction = TfidfVectorizer(min_df = 1, stop_words='english', lowercase=True)
# min_df = 1 ----- Ignore words that appear in fewer than 1 document.
# stop_words='english' ----- Removes common English words..
# lowercase=True ---   Converts every word into lowercase
```

```python
X_test_features = feature_extraction.fit_transform(X_test)
X_train_features = feature_extraction.transform(X_train)

y_train_features = Y_train.astype('int')
y_test_features = Y_test.astype('int')
```

```python
print(X_train_features , X_test_features)
print(y_train_features , y_test_features)
```

train ML model

```python
model = LogisticRegression()
```

```python
model.fit(X_train_features,y_train_features )
```

evalute the training model

```python
pred_on_training_data = model.predict(X_train_features)
accuracy_on_training_data = accuracy_score(y_train_features, pred_on_training_data)
```

```python
print('Accuracy on training data : ', accuracy_on_training_data)
```

```python
pred_on_test_data = model.predict(X_test_features)
accuracy_on_test_data = accuracy_score(y_test_features, pred_on_test_data)
```

```python
print('Accuracy on test data : ', accuracy_on_test_data)
```
