```python
### What is the average value of each numerical column for each group?
df.groupby(60).mean()
```

```python
# separating based on data and label
# There are two axes:
# axis=0 → Rows (vertical)
# axis=1 → Columns (horizontal)

X = df.drop(columns=60, axis=1)
Y = df[60]
```

```python
print(X)  # dataframe only with features
print(Y)  # data only with lable
```

training and test data

```python
# straitify=Y ,This ensures the class proportions are maintained in both the training and testing sets.
X_train , x_test , Y_train , y_test = train_test_split(X,Y,test_size=0.1,stratify=Y,random_state=42)
```

```python
print(X.shape,X_train.shape,x_test.shape)
```

```python
print(Y.shape,Y_train.shape,y_test.shape)
```

```python
print(X_train)
print(Y_train)
```

model training -> logistic regresion


```python
model = LogisticRegression()
```

```python
# training the logistic Regression model with training data
# "Learn the relationship between the input features (X_train) and the correct answers (Y_train)."
model.fit(X_train , Y_train)
```

Model Evaluation

```python
#accuracy on training data
X_train_prediction = model.predict(X_train)
training_data_accuracy = accuracy_score(X_train_prediction,Y_train )
```

```python
print('Accuracy on training : ', training_data_accuracy)
```

```python
#accuracy on testing data
X_testing_prediction = model.predict(x_test)
testing_data_accuracy = accuracy_score(X_testing_prediction,y_test )
```

```python
print('Accuracy on test data : ', testing_data_accuracy)
```
